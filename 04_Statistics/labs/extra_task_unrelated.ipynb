```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sukuriame bandomuosius duomenis tavo projektui
data = {
    'Darbuotojas': ['Jonas', 'Asta', 'Marius', 'Lina', 'Tomas'],
    'Valandos': [38, 42, 35, 40, 39]
}
df = pd.DataFrame(data)
```

```python
#Matplotlib

plt.figure(figsize=(8, 5))
plt.bar(df['Darbuotojas'], df['Valandos'], color='skyblue')
plt.title('Darbo valandos per savaitę (Matplotlib)')
plt.xlabel('Darbuotojas')
plt.ylabel('Valandos')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def generuoti_atlyginimu_ataskaita(darbuotojai, valandos, ikainiai):
    # 1. Sukuriame DataFrame (Pandas dalis)
    df = pd.DataFrame({
        'Darbuotojas': darbuotojai,
        'Dirbtos valandos': valandos,
        'Valandinis įkainis': ikainiai
    })
    
    # 2. Apskaičiuojame atlyginimą (Payroll logika)
    df['Atlyginimas'] = df['Dirbtos valandos'] * df['Valandinis įkainis']
    
    # 3. Vizualizacija (Seaborn dalis)
    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(10, 6))
    
    # Braižome stulpelinę diagramą
    plot = sns.barplot(
        data=df, 
        x='Darbuotojas', 
        y='Atlyginimas', 
        palette='magma',
        hue='Darbuotojas',
        legend=False
    )
    
    # Pridedame tekstines vertes virš stulpelių (Matplotlib derinimas)
    for p in plot.patches:
        plot.annotate(format(p.get_height(), '.2f') + ' €', 
                      (p.get_x() + p.get_width() / 2., p.get_height()), 
                      ha = 'center', va = 'center', 
                      xytext = (0, 9), 
                      textcoords = 'offset points')

    plt.title('Savaitės biudžetas: Atlyginimai išmokėjimui', fontsize=14)
    plt.ylabel('Suma (€)')
    plt.xlabel('Darbuotojas')
    plt.ylim(0, df['Atlyginimas'].max() * 1.2) # Paliekame vietos tekstui viršuje
    
    plt.show()
    return df

# Testuojame funkciją su tavo 5 darbuotojais
darbuotojai = ['Jonas', 'Asta', 'Marius', 'Lina', 'Tomas']
valandos = [40, 38, 45, 20, 42]
ikainiai = [12.5, 12.5, 15.0, 11.0, 13.0]

rezultatai = generuoti_atlyginimu_ataskaita(darbuotojai, valandos, ikainiai)
print(rezultatai)
```
