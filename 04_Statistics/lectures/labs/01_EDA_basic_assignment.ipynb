## Assignment 2: Variable Relationships

1. Build a correlation matrix of numeric features and pass it into a heatmap
2. Build a pairplot of numeric features in the data. Hint: df.select_dtypes("number") selects numeric columns. Consider sampling with df.sample(1000) to speed this up if it takes too long.
3. Build an lmplot of ram vs. price. Does being a premium model (premium == 'yes') impact this relationship? 

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Takes only numeric variables
num = df.select_dtypes("number")
```

```python
# 1. Build a correlation matrix of numeric features and pass it into a heathmap
## Calculates the correlation between all the numeric variables
corr = num.corr()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap="vlag")
plt.show()
```

```python
## 2. Build a pairplot of numeric features

sns.pairplot(num)
plt.show()
```

```python
sns.pairplot(num.sample(min(1000, len(num))))
plt.show()
```

```python
#3. lmplot ram vs. price, premium as hue. Linear regression model. 
# Does premium significantly factors the price..?
sns.lmplot(data=df, x="ram", y="price", hue="premium")
plt.show()
```
