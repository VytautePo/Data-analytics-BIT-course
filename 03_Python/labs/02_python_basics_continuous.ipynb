1. Create a list with numbers from 1 to 10. Use a for loop and print only the even numbers. Use continue.

```python
numbers = list(range(1, 11))
for n in numbers:
    if n % 2 == 0:
        continue
    print(n)
```

2. You have a list: ['apple', 'banana', 'cherry', 'date']. Use a for loop and if statement to print the fruits whose length is greater than 5.

```python
fruit = ['apple', 'banana', 'cherry', 'date']
for f in fruit:
    if len(f) > 5:
        print(f)
```

3. Using a while loop, calculate the sum of numbers from 1 to 20. If the sum exceeds 100, exit the loop using break

```python
i = 1
total = 0

while i <= 20:
    total += i
    if total > 100:
        break
    i += 1
    
print(total)
```

4. You have a dictionary with students' grades. Print only the students whose grade is greater than or equal to 8. Use if and for.

```python
students = {'Ryzis': 9, 'Ceske': 7, 'Dzeris': 8}
for name, grade in students.items():
    if grade >= 8:
        print(name, grade)
```

7. Use a for loop to double each value in the list [2, 4, 6]. The result should be a new list.

```python
numbers = [2, 4, 6]
doubled = []

for n in numbers:
    doubled.append(n * 2)
    
print(doubled)
```

```python
# more popular and practical list comprehension option

numbers = [2, 5, 3, 8, 1, 6]

numbers = [n for n in numbers if n >= 5]

print(numbers)  ##doesn't remove anyting, but creates a new list with only values bigger than 5
```

```python
### extra practice with library comprehension 
grades = {
    "Jonas": 7,
    "Ona": 9,
    "Petras": 5,
    "Ieva": 10
}

grades = {name: grade + 1 for name, grade in grades.items() if grade >= 8}
print(grades)
```
