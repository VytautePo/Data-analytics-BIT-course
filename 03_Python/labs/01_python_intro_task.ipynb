5. Calculate the price without VAT (121 € including 21% VAT).

```python
price = 121
vat = 121 * 0.21
price_wo_vat = price - vat
print("Price without VAT:", price_wo_vat, "EUR")
```

```python
price_with_vat = 121
vat_rate = 0.21

price_without_vat = price_with_vat / (1 + vat_rate)

print("Price without VAT:", round(price_without_vat, 2), "EUR")
```

6. Calculate the profit after one year with 5% interest on 1000 €.

```python
base = 1000
interest = 0.05
year_profit = base * interest
total_amount = base + year_profit
print("Profit after one year:", year_profit, "EUR")
print("Total amount with interest:", total_amount, "EUR")
```

7. Calculate the average of the grades [7, 8, 9].

8. Calculate what percentage of the loan remains unpaid (1000 € total – 350 € repaid).

```python
total = 1000
repaid = 350
unpaid = total - repaid
unpaid_percentage = (unpaid / total) * 100
print("Percentage of the loan to be paid yet:", unpaid_percentage, "%")
```

19. Extract the year, month, and day from the date "2025-06-01".

```python
# with string split
date = "2025-06-01"
year, month, day = date.split("-")
print("Year:", year)
print("Month:", month)
print("Day:", day)
```

```python
# with datetime if we need to use date for some calculations 
from datetime import datetime
date = "2025-06-01"
dt = datetime.strptime(date, "%Y-%m-%d")
print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
```

23. Check if a number is even or odd.

```python
# We use modular operator %
# to use it for even % 2 == 0: % retuns what is left after division, in this case division form 2, if none left - even
# to search first for odd % 2 != 0: if some value is left and not 0 after the division from 2, the number is odd
number = 9
if number % 2 == 0:
    print("Number is even")
else:
    print("Number is odd")
```

25. Input temperature – if >25 → "hot", 10–25 → "normal", otherwise → "cold".
