3. Užduotis: Algos funkcija Parašyk funkciją calculate_pay(hours, rate). Ji turi gauti išdirbtas valandas ir tarifą, bei grąžinti (return) galutinę sumą. Patikrink funkciją su skaičiais: 40 valandų ir 15 eurų tarifas.

```python
def calculate_pay(hours, rate):
    return hours * rate

print("Employee total:", calculate_pay(40, 15), "EUR" )
```

4. Užduotis: Viršvalandžių detektorius Parašyk funkciją is_overtime(hours), kuri grąžina True, jei valandų skaičius yra daugiau nei 40, ir False visais kitais atvejais.

```python
def is_overtime(hours):
    return hours > 40

print("Worked oovertime:", is_overtime(16))
```

5. Užduotis: Komandos valandų suma Turi sąrašą su visų 5 darbuotojų išdirbtomis valandomis: total_hours = [40, 45, 38, 40, 42]. Naudodama sum() ir len(), apskaičiuok ir atspausdink, koks yra tavo komandos valandų vidurkis.

```python
total_hours = [40, 45, 38, 35]

def avg_team_hors(total_hours):
    return sum(total_hours) / len(total_hours)
result = avg_team_hors(total_hours)

print(f"Average team hours: {result}")
```

6. Tavo užduotis: „Algalapio“ kūrimas
Įsivaizduok, kad visi darbuotojai šią savaitę išdirbo po 40 valandų, bet Eglė pasidarbavo daugiau ir išdirbo 45 valandas.

```python
workers_rate = {'Carmen': 25, 'Egle': 33, 'Jorge': 20, 'Alessia': 18}
worked_hours = {'Carmen': 40, 'Egle': 45, 'Jorge': 40, 'Alessia': 40}

for name in workers_rate:
    rate = workers_rate[name]
    hours = worked_hours[name]
    
    total_salary = calculate_pay(hours, rate)
    overtime_status = is_overtime(hours)
    status_text = "DIRBO" if overtime_status else "nedirbo"
    
    print(f"Darbuotojas: {name} | Alga: {total_salary} EUR | Virsvalandziai: {status_text}")
```

Su total payroll:

```python
workers_rate = {'Carmen': 25, 'Egle': 33, 'Jorge': 20, 'Alessia': 18}
worked_hours = {'Carmen': 40, 'Egle': 45, 'Jorge': 40, 'Alessia': 40}
total_payroll = 0

for name in workers_rate:
    rate = workers_rate[name]
    hours = worked_hours.get(name, 0)
    
    salary = calculate_pay(hours, rate)
    total_payroll += salary
    
    overtime_status = is_overtime(hours)
    status_text = "DIRBO" if overtime_status else "nedirbo"
    
    print(f"Darbuotojas: {name:10} | Alga: {salary:4} EUR | Virsvalandziai: {status_text}")
    
print("-" * 70)
print(f"IS VISO MOKETI: {total_payroll} EUR")
```
