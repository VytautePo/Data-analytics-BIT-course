```python
pradzia = datetime.date(2024, 6, 1)
pabaiga = datetime.date(2024, 6, 17)

diena = pradzia
while diena <= pabaiga:
    print(diena)
    diena += datetime.timedelta(days=1)
```

```python
gimimo_data = datetime.date(1989, 5, 26)
siandien = datetime.date.today()

amzius = siandien.year - gimimo_data.year
if (siandien.month, siandien.day) < (gimimo_data.month, gimimo_data.day):
    amzius -= 1

print("Amžius:", amzius)
```

```python
import locale
from datetime import datetime

locale.setlocale(locale.LC_TIME, 'lt_LT.UTF-8')
now = datetime.now()
day_name = now.strftime("%A")
print(day_name)  
```

```python
import locale
from datetime import datetime

locale.setlocale(locale.LC_TIME, 'lt_LT.UTF-8')
now = datetime.now()
day_name = now.strftime("%A")
year = now.strftime("%Y")
#print("Siandien yra", day_name, ",", year, "metai")
print(f"Siandien yra {day_name}, {year} metai")

```
